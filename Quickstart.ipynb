# 快速开始-1：编程建模中调用单模型

1. 引入`pygeomodels`包，并初始化模型库

```python
from pygeomodels.config import parse_config
from pygeomodels.modelBank import modelBank
from pygeomodels.modelTask import modelTask
```

```python
cfg = parse_config()
model_bank = modelBank(cfg)
caller = model_bank.models_caller
```

2. 浏览、预处理数据，为调用模型做准备

```python
import os
dem_dir= '/onesis/kt4/test_data/terr'
# dem_files = [f for f in os.listdir(dem_dir) if os.path.isfile(os.path.join(dem_dir, f))]
# print(dem_files)
out_dir = '/onesis/kt4/job_results'
# if not os.path.exists(out_dir):
#     os.mkdir(out_dir)
```

3. 确定好输入数据、模型参数、输出数据后，调用模型（即提交模型计算任务）

```python
# 以调用DEM填洼算法为例
inputs = {"z": dem_dir + "/dem_meixi.tif"}
params = {}
outputs = {"fel": out_dir + "/dem_meixi_fel_notebook.tif"}

task_id = caller.pitRemove(inputs, params, outputs)
```

4. 查看模型计算任务是否提交成功

```python
if task_id is not None:
    print('Task submitted succeed, task ID: %s' % task_id)
else:
    print('Task submitted failed!')
```

5. 查看模型计算相关数据信息、运行日志、任务状态，以及删除任务等

```python
task = modelTask(cfg, task_id)
print(task.data_info())
```

```python
print(task.log())
```

```python
print(task.progress())
```

```python
print(task.delete())
```
